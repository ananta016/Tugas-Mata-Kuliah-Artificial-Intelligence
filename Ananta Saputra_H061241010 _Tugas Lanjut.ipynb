### **1. STUDI KASUS GEOLISTRIK**

### **Deskripsi**

Kajian ini memakai keluaran survei **geolistrik**, seperti citra penampang resistivitas dua dimensi maupun data sounding.

Pembacaan data diarahkan untuk menemukan perubahan sifat listrik di bawah permukaan sehingga susunan lapisan, perbedaan material, struktur, atau zona tertentu dapat dikenali.

### **Role**

> Anda bertindak sebagai pendamping analisis yang membantu menerangkan informasi yang ditunjukkan oleh data geolistrik.

### **Task**

> Membantu membaca dan menafsirkan pola resistivitas yang tersedia sehingga perubahan karakter bawah permukaan dapat dijelaskan secara geofisika.

### **Context**

> Geolistrik bekerja dengan memanfaatkan perbedaan respons listrik material di bawah tanah. Resistivitas suatu material dapat berubah karena jenis material, porositas, kandungan air atau fluida, kejenuhan, dan lingkungan geologinya.

### **Input**

> Hasil yang digunakan berupa penampang resistivitas, angka resistivitas, informasi kedalaman, susunan atau posisi elektroda, serta keterangan geologi maupun hidrogeologi apabila tersedia.

### **Constraints**

> Tidak melakukan perhitungan ulang maupun menambahkan angka yang tidak terdapat pada data. Penjelasan harus bersumber dari informasi yang tersedia dan memisahkan fakta pengukuran, hasil penafsiran, serta kemungkinan interpretasi.

### **Output**

> Paparkan kecenderungan resistivitas, pembagian lapisan atau zona, perubahan arah mendatar dan vertikal, serta dugaan karakter material yang dapat menjelaskan pola tersebut.

### **Quality Criteria**

> Penafsiran harus mempertimbangkan legenda, rentang warna, besaran resistivitas, kedalaman, dan data pendukung. Istilah geofisika digunakan sesuai maknanya dan hasil dijelaskan secara runtut.

### **Verification**

> Tinjau kembali apakah uraian yang dibuat benar-benar ditunjukkan oleh penampang resistivitas. Jika ada data pendukung, lakukan pencocokan dengan informasi geologi, sumur atau bor, maupun hasil survei lainnya.


## Eksekusi 1: Geolistrik

**Data sintetis:** lintasan 100 m, kedalaman model 20 m, dan zona resistivitas rendah di bagian tengah penampang.

Kode dan visualisasi berikut menggunakan data sintetis untuk demonstrasi.

**Tujuan eksekusi:** menghasilkan keluaran terukur dan visual sebagai bahan untuk menyusun contoh interpretasi berdasarkan prompt asli di atas.

```python

import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from IPython.display import display, Image

# ============================================================
# 1. PEMBUATAN MODEL RESISTIVITAS SINTETIS
# ============================================================

x = np.linspace(0, 100, 101)   # Jarak lintasan (m)
z = np.linspace(0, 20, 41)     # Kedalaman (m)
X, Z = np.meshgrid(x, z)

# Resistivitas latar belakang
rho = np.full_like(X, 60.0)

# Anomali resistivitas rendah
rho[(X >= 35) & (X <= 65) & (Z >= 4) & (Z <= 13)] = 14.0

# Anomali resistivitas tinggi
rho[(X >= 72) & (Z >= 8)] = 220.0

# ============================================================
# 2. INFORMASI MODEL
# ============================================================

print("=" * 55)
print("METODE GEOLISTRIK - MODEL RESISTIVITAS SINTETIS")
print("=" * 55)
print(f"Resistivitas minimum : {rho.min():.1f} ohm.m")
print(f"Resistivitas maksimum: {rho.max():.1f} ohm.m")
print("Panjang lintasan     : 100 m")
print("Kedalaman model      : 20 m")
print("=" * 55)

# ============================================================
# 3. VISUALISASI PENAMPANG RESISTIVITAS
# ============================================================

fig, ax = plt.subplots(figsize=(11, 5))

im = ax.pcolormesh(
    X, Z, rho,
    shading="auto",
    cmap="viridis",
    vmin=0,
    vmax=220
)

ax.invert_yaxis()

cbar = fig.colorbar(im, ax=ax, pad=0.02)
cbar.set_label("Resistivitas (ohm.m)")

ax.set_xlabel("Jarak lintasan (m)")
ax.set_ylabel("Kedalaman (m)")
ax.set_title("Penampang Resistivitas Sintetis")

ax.set_xticks(np.arange(0, 101, 10))
ax.set_yticks(np.arange(0, 21, 2))
ax.grid(False)

fig.tight_layout()

# ============================================================
# 4. SIMPAN DAN TAMPILKAN GRAFIK
# ============================================================

nama_file = "penampang_resistivitas.png"
fig.savefig(nama_file, dpi=200, bbox_inches="tight")
plt.close(fig)

display(Image(filename=nama_file))

print(f"\nGrafik berhasil disimpan: {nama_file}")
```

Zona resistivitas rendah pada jarak sekitar 35–65 m dan kedalaman 4–13 m dapat berkaitan dengan material lebih konduktif, misalnya tanah lempungan atau material jenuh air. Kedua kemungkinan tersebut tidak dapat dibedakan hanya dari resistivitas. Zona resistivitas tinggi di bagian kanan dapat mengindikasikan material lebih kompak atau kering. Verifikasi membutuhkan data geologi atau pengeboran.

## Eksekusi 2: Seismik Refraksi

**Data sintetis:** 24 geofon dengan jarak antargeofon 5 m; waktu tiba dibentuk dari model dua lapisan untuk menunjukkan tahapan analisis.

Kode dan visualisasi berikut menggunakan data sintetis untuk demonstrasi.

**Tujuan eksekusi:** menghasilkan keluaran terukur dan visual sebagai bahan untuk menyusun contoh interpretasi berdasarkan prompt asli di atas.

```python

import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from IPython.display import display, Image

# ============================================================
# 1. PEMBUATAN DATA SEISMIK REFRAKSI SINTETIS
# ============================================================

x = np.arange(0, 116, 5, dtype=float)

# Parameter model dua lapisan
v1 = 500.0    # Kecepatan lapisan 1 (m/s)
v2 = 1800.0   # Kecepatan lapisan 2 (m/s)
ti = 0.030    # Waktu intersep (s)

# Waktu tempuh gelombang
t_direct = x / v1
t_head = ti + x / v2
t = np.minimum(t_direct, t_head)

# Konversi detik ke milidetik
t_ms = t * 1000
t_direct_ms = t_direct * 1000
t_head_ms = t_head * 1000

# ============================================================
# 2. INFORMASI DATA
# ============================================================

print("=" * 55)
print("SEISMIK REFRAKSI - DATA SINTETIS")
print("=" * 55)
print(f"Jumlah posisi penerima : {len(x)}")
print(f"Jarak antar penerima   : {x[1] - x[0]:.1f} m")
print(f"Kecepatan lapisan 1    : {v1:.1f} m/s")
print(f"Kecepatan lapisan 2    : {v2:.1f} m/s")
print(f"Waktu intersep         : {ti * 1000:.1f} ms")
print(f"Waktu tiba minimum     : {t_ms.min():.2f} ms")
print(f"Waktu tiba maksimum    : {t_ms.max():.2f} ms")
print("Parameter merupakan model sintetis.")
print("=" * 55)

# ============================================================
# 3. VISUALISASI KURVA WAKTU TIBA
# ============================================================

fig, ax = plt.subplots(figsize=(9, 5))

ax.plot(
    x, t_ms,
    "o-",
    linewidth=1.8,
    markersize=5,
    label="Waktu tiba sintetis"
)

ax.plot(
    x, t_direct_ms,
    "--",
    linewidth=1.5,
    label="Gelombang langsung"
)

ax.plot(
    x, t_head_ms,
    "--",
    linewidth=1.5,
    label="Gelombang refraksi"
)

ax.set_xlabel("Jarak sumber–penerima (m)")
ax.set_ylabel("Waktu tiba (ms)")
ax.set_title("Kurva Waktu Tiba Seismik Refraksi Sintetis")
ax.grid(True, alpha=0.3)
ax.legend()

fig.tight_layout()

# ============================================================
# 4. SIMPAN DAN TAMPILKAN GRAFIK
# ============================================================

nama_file = "kurva_seismik_refraksi.png"

fig.savefig(nama_file, dpi=200, bbox_inches="tight")
plt.close(fig)

display(Image(filename=nama_file))

print(f"\nGrafik berhasil disimpan: {nama_file}")
```

Kurva waktu tiba memperlihatkan perubahan kemiringan yang dalam model refraksi sederhana berkaitan dengan perubahan kecepatan rambat. Kecepatan lapisan yang lebih dalam pada contoh ini ditetapkan lebih tinggi daripada lapisan atas. Karena data dibangkitkan dari parameter model, hasil ini hanya menunjukkan prinsip analisis, bukan estimasi kecepatan lapangan.

## Eksekusi 3: Ground Penetrating Radar (GPR)

**Data sintetis:** frekuensi antena 200 MHz, kecepatan gelombang 0,10 m/ns, lintasan 0–30 m, dan jendela waktu 80 ns.

Kode dan visualisasi berikut menggunakan data sintetis untuk demonstrasi.

**Tujuan eksekusi:** menghasilkan keluaran terukur dan visual sebagai bahan untuk menyusun contoh interpretasi berdasarkan prompt asli di atas.

```python

import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from IPython.display import display, Image

# ============================================================
# 1. PEMBUATAN MODEL RADARGRAM GPR SINTETIS
# ============================================================

x = np.linspace(0, 30, 241)       # Jarak lintasan (m)
t_ns = np.linspace(0, 80, 201)    # Waktu tempuh dua arah (ns)

X, T = np.meshgrid(x, t_ns)

# Membuat noise sintetis
rng = np.random.default_rng(7)
radar = 0.12 * rng.normal(size=X.shape)

# Refleksi horizontal sintetis
radar += (
    0.9 * np.exp(-((T - 18) / 2.2)**2)
    * np.cos(2 * np.pi * (T - 18) / 5)
)

radar += (
    0.65 * np.exp(-((T - 42) / 3.0)**2)
    * np.cos(2 * np.pi * (T - 42) / 6)
)

# ============================================================
# 2. PEMODELAN ANOMALI HIPERBOLA
# ============================================================

v = 0.10        # Kecepatan gelombang (m/ns)
x0 = 14.0       # Posisi lateral anomali (m)
z0 = 1.2        # Kedalaman target sintetis (m)

# Waktu tempuh dua arah pada posisi target
t0 = 2 * z0 / v

# Kurva hiperbola waktu tempuh
hyper_t = np.sqrt(
    t0**2 + (2 * (X - x0) / v)**2
)

# Menambahkan respons hiperbola sintetis
radar += 1.1 * np.exp(-((T - hyper_t) / 1.8)**2)

# ============================================================
# 3. INFORMASI MODEL
# ============================================================

time_window = 80.0
zmax = v * time_window / 2

print("=" * 55)
print("GPR - RADARGRAM DATA SINTETIS")
print("=" * 55)
print(f"Panjang lintasan          : {x.max():.1f} m")
print(f"Time window               : {time_window:.1f} ns")
print(f"Kecepatan asumsi          : {v:.2f} m/ns")
print(f"Kedalaman maksimum teoritis: {zmax:.2f} m")
print(f"Posisi anomali            : {x0:.1f} m")
print(f"Kedalaman target sintetis : {z0:.1f} m")
print("=" * 55)

# ============================================================
# 4. VISUALISASI RADARGRAM
# ============================================================

fig, ax = plt.subplots(figsize=(11, 5))

im = ax.pcolormesh(
    X, T, radar,
    shading="auto",
    cmap="gray"
)

ax.invert_yaxis()

fig.colorbar(im, ax=ax, label="Amplitudo relatif")
ax.set_xlabel("Jarak lintasan (m)")
ax.set_ylabel("Waktu tempuh dua arah (ns)")
ax.set_title("Radargram GPR Sintetis")

fig.tight_layout()

# ============================================================
# 5. SIMPAN DAN TAMPILKAN GRAFIK
# ============================================================

nama_file = "radargram_gpr_sintetis.png"

fig.savefig(nama_file, dpi=200, bbox_inches="tight")
plt.close(fig)

display(Image(filename=nama_file))

print(f"\nRadargram berhasil disimpan: {nama_file}")
```

Radargram menampilkan refleksi berlapis dan anomali berbentuk hiperbola di sekitar posisi lintasan 14 m. Bentuk hiperbola dapat berkaitan dengan objek lokal atau perubahan material, tetapi tidak cukup untuk menetapkan bahwa objek tersebut adalah pipa atau benda tertentu. Dengan kecepatan asumsi 0,10 m/ns dan waktu 80 ns, kedalaman maksimum teoritis adalah 4 m; kedalaman efektif nyata dapat lebih kecil akibat atenuasi.

## Eksekusi 4: Geomagnet

```python

import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from IPython.display import display, Image

# ============================================================
# 1. MEMBUAT GRID DAN ANOMALI MAGNETIK SINTETIS
# ============================================================

x = np.linspace(0, 400, 41)
y = np.linspace(0, 400, 41)
X, Y = np.meshgrid(x, y)

# Pusat sumber sintetis
x0, y0 = 220.0, 245.0

# Jarak kuadrat dari pusat sumber
r2 = (X - x0)**2 + (Y - y0)**2

# Model anomali magnetik sintetis
anomaly = (
    280 * np.exp(-r2 / (2 * 65.0**2))
    - 75 * np.exp(-r2 / (2 * 130.0**2))
)

# ============================================================
# 2. INFORMASI DATA
# ============================================================

print("=" * 55)
print("GEOMAGNET - DATA SINTETIS")
print("=" * 55)
print(f"Anomali minimum : {anomaly.min():.1f} nT")
print(f"Anomali maksimum: {anomaly.max():.1f} nT")
print(f"Pusat model X   : {x0:.0f} m")
print(f"Pusat model Y   : {y0:.0f} m")
print("Data ini merupakan simulasi, bukan data lapangan.")
print("=" * 55)

# ============================================================
# 3. VISUALISASI PETA KONTUR ANOMALI MAGNETIK
# ============================================================

fig, ax = plt.subplots(figsize=(7.5, 6))

cf = ax.contourf(
    X, Y, anomaly,
    levels=20,
    cmap="RdBu_r"
)

ax.contour(
    X, Y, anomaly,
    levels=10,
    colors="black",
    linewidths=0.45,
    alpha=0.6
)

ax.plot(
    x0, y0,
    "k+",
    markersize=12,
    markeredgewidth=2,
    label="Pusat sumber model"
)

fig.colorbar(
    cf,
    ax=ax,
    label="Anomali magnetik sintetis (nT)"
)

ax.set_xlabel("X (m)")
ax.set_ylabel("Y (m)")
ax.set_title("Peta Kontur Anomali Geomagnet Sintetis")
ax.set_aspect("equal", adjustable="box")
ax.legend()

fig.tight_layout()

# ============================================================
# 4. SIMPAN DAN TAMPILKAN PETA
# ============================================================

nama_file = "peta_geomagnet_sintetis.png"

fig.savefig(nama_file, dpi=200, bbox_inches="tight")
plt.close(fig)

display(Image(filename=nama_file))

print(f"\nPeta berhasil disimpan: {nama_file}")
```

Peta sintetis memperlihatkan anomali positif yang terpusat dan melemah menjauhi pusatnya, dengan komponen negatif lemah di area lebih luas. Pola seperti ini dapat digunakan untuk mendiskusikan kemungkinan kontras sifat magnetik bawah permukaan. Pada survei nyata, interpretasi sumber harus mempertimbangkan koreksi IGRF/variasi harian, arah medan magnet, kedalaman, bentuk sumber, dan kondisi geologi.

## Eksekusi 5: Induced Polarization (IP)

**Data sintetis:** lintasan 0–100 m, kedalaman 0–25 m, serta zona chargeability tinggi dan resistivitas relatif rendah di bagian tengah.

Kode dan visualisasi berikut menggunakan data sintetis untuk demonstrasi.

**Tujuan eksekusi:** menghasilkan keluaran terukur dan visual sebagai bahan untuk menyusun contoh interpretasi berdasarkan prompt asli di atas.

```python

import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from IPython.display import display, Image

# ============================================================
# 1. PEMBUATAN MODEL IP SINTETIS
# ============================================================

x = np.linspace(0, 100, 101)  # Jarak lintasan (m)
z = np.linspace(0, 25, 51)    # Kedalaman (m)
X, Z = np.meshgrid(x, z)

# Model latar belakang
rho = np.full_like(X, 80.0)  # Resistivitas (ohm.m)
m = np.full_like(X, 8.0)     # Chargeability (mV/V)

# Zona anomali sintetis
target = (
    (X >= 40) & (X <= 65) &
    (Z >= 6) & (Z <= 18)
)

rho[target] = 28.0
m[target] = 32.0

# ============================================================
# 2. INFORMASI MODEL
# ============================================================

print("=" * 58)
print("INDUCED POLARIZATION (IP) - DATA SINTETIS")
print("=" * 58)
print(f"Rentang resistivitas : {rho.min():.1f}–{rho.max():.1f} ohm.m")
print(f"Rentang chargeability: {m.min():.1f}–{m.max():.1f} mV/V")
print("Zona anomali         : X = 40–65 m")
print("Kedalaman anomali    : 6–18 m")
print("Data merupakan simulasi, bukan data lapangan.")
print("=" * 58)

# ============================================================
# 3. VISUALISASI RESISTIVITAS DAN CHARGEABILITY
# ============================================================

fig, axes = plt.subplots(
    1, 2,
    figsize=(12, 5),
    sharey=True
)

# Penampang resistivitas
im1 = axes[0].pcolormesh(
    X, Z, rho,
    shading="auto",
    cmap="viridis"
)

axes[0].set_title("Resistivitas")
axes[0].set_xlabel("Jarak lintasan (m)")
axes[0].set_ylabel("Kedalaman (m)")

fig.colorbar(
    im1,
    ax=axes[0],
    label="Resistivitas (ohm.m)"
)

# Penampang chargeability
im2 = axes[1].pcolormesh(
    X, Z, m,
    shading="auto",
    cmap="plasma"
)

axes[1].set_title("Chargeability")
axes[1].set_xlabel("Jarak lintasan (m)")

fig.colorbar(
    im2,
    ax=axes[1],
    label="Chargeability (mV/V)"
)

# Kedalaman bertambah ke bawah
for ax in axes:
    ax.invert_yaxis()
    ax.set_xticks(np.arange(0, 101, 10))
    ax.set_yticks(np.arange(0, 26, 5))

fig.suptitle("Penampang IP Sintetis", fontsize=14)
fig.tight_layout()

# ============================================================
# 4. SIMPAN DAN TAMPILKAN GRAFIK
# ============================================================

nama_file = "penampang_ip_sintetis.png"

fig.savefig(
    nama_file,
    dpi=200,
    bbox_inches="tight"
)

plt.close(fig)

display(Image(filename=nama_file))

print(f"\nGrafik berhasil disimpan: {nama_file}")
```

Zona model pada jarak 40–65 m dan kedalaman 6–18 m memiliki resistivitas lebih rendah serta chargeability lebih tinggi dibandingkan latar belakang. Kombinasi ini menunjukkan respons listrik yang berbeda, tetapi penyebabnya belum dapat ditentukan secara unik. Mineralisasi, material lempungan, dan faktor geologi lain perlu dievaluasi menggunakan data pendukung.
