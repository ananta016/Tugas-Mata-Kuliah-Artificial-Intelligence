### **1. STUDI KASUS GEOLISTRIK**

### **Deskripsi**

Kajian ini memakai keluaran survei **geolistrik**, seperti citra penampang resistivitas dua dimensi maupun data sounding.

Pembacaan data diarahkan untuk menemukan perubahan sifat listrik di bawah permukaan sehingga susunan lapisan, perbedaan material, struktur, atau zona tertentu dapat dikenali.

### **Role**

> Anda bertindak sebagai pendamping analisis yang membantu menerangkan informasi yang ditunjukkan oleh data geolistrik.

### **Task**

> Membantu membaca dan menafsirkan pola resistivitas yang tersedia sehingga perubahan karakter bawah permukaan dapat dijelaskan secara geofisika.

### **Context**

> Geolistrik bekerja dengan memanfaatkan perbedaan respons listrik material di bawah tanah. Resistivitas suatu material dapat berubah karena jenis material, porositas, kandungan air atau fluida, kejenuhan, dan lingkungan geologinya.

### **Input**

> Hasil yang digunakan berupa penampang resistivitas, angka resistivitas, informasi kedalaman, susunan atau posisi elektroda, serta keterangan geologi maupun hidrogeologi apabila tersedia.

### **Constraints**

> Tidak melakukan perhitungan ulang maupun menambahkan angka yang tidak terdapat pada data. Penjelasan harus bersumber dari informasi yang tersedia dan memisahkan fakta pengukuran, hasil penafsiran, serta kemungkinan interpretasi.

### **Output**

> Paparkan kecenderungan resistivitas, pembagian lapisan atau zona, perubahan arah mendatar dan vertikal, serta dugaan karakter material yang dapat menjelaskan pola tersebut.

### **Quality Criteria**

> Penafsiran harus mempertimbangkan legenda, rentang warna, besaran resistivitas, kedalaman, dan data pendukung. Istilah geofisika digunakan sesuai maknanya dan hasil dijelaskan secara runtut.

### **Verification**

> Tinjau kembali apakah uraian yang dibuat benar-benar ditunjukkan oleh penampang resistivitas. Jika ada data pendukung, lakukan pencocokan dengan informasi geologi, sumur atau bor, maupun hasil survei lainnya.


### **2. STUDI KASUS SEISMIK**

### **Deskripsi**

Kajian ini menggunakan keluaran survei **seismik**, misalnya rekaman atau penampang seismik, data waktu tiba, dan model kecepatan gelombang.

Analisis diarahkan untuk membaca perubahan respons gelombang yang dapat menunjukkan batas lapisan, variasi kecepatan, serta bentuk struktur bawah permukaan.

### **Role**

> Anda bertindak sebagai pendamping yang membantu menguraikan informasi bawah permukaan berdasarkan hasil pengukuran seismik.

### **Task**

> Membantu membaca karakter refleksi atau refraksi serta perubahan kecepatan agar susunan dan struktur bawah permukaan dapat dijelaskan dari data yang tersedia.

### **Context**

> Survei seismik memanfaatkan penjalaran gelombang elastik melalui material bumi. Respons gelombang dipengaruhi oleh sifat fisik medium, seperti kecepatan, densitas, kekompakan, porositas, dan perubahan struktur.

### **Input**

> Data masukan dapat berupa penampang seismik, waktu tiba, parameter kecepatan, koordinat sumber dan penerima, serta keterangan geologi yang mendukung interpretasi.

### **Constraints**

> Jangan melakukan rekonstruksi data atau memasukkan parameter yang tidak tersedia. Setiap kesimpulan harus ditelusuri dari data dan harus dibedakan antara pengamatan langsung, penafsiran, dan kemungkinan yang belum terkonfirmasi.

### **Output**

> Terangkan posisi dan pola reflektor atau refraktor, perubahan kecepatan, batas antarunit, bentuk struktur, serta karakter material yang mungkin berkaitan dengan respons seismik.

### **Quality Criteria**

> Hasil pembacaan harus mengacu pada skala, waktu atau kedalaman, legenda, nilai kecepatan, dan informasi pendukung. Gunakan istilah seismik secara tepat dan hindari interpretasi yang melampaui data.

### **Verification**

> Cek kembali hubungan antara interpretasi dan data seismik. Apabila tersedia, gunakan data geologi, sumur, atau pengukuran lain sebagai pembanding.


### **3. STUDI KASUS GPR**

### **Deskripsi**

Kajian ini memanfaatkan hasil **Ground Penetrating Radar (GPR)** dalam bentuk radargram atau penampang bawah permukaan yang telah diperoleh dari survei.

Perhatian utama diberikan pada bentuk dan kesinambungan reflektor elektromagnetik untuk mengenali batas material, objek dangkal, lapisan, maupun perubahan kondisi bawah permukaan.

### **Role**

> Anda bertindak sebagai pendamping yang membantu membaca serta menjelaskan informasi yang terdapat pada hasil survei GPR.

### **Task**

> Membantu menguraikan radargram dengan melihat karakter reflektor, posisi, kontinuitas, dan perubahan pola untuk mengenali kondisi bawah permukaan.

### **Context**

> GPR menggunakan pulsa elektromagnetik berfrekuensi tinggi yang merambat ke dalam tanah dan dipantulkan ketika menemui perubahan sifat material. Responsnya berkaitan dengan permitivitas, konduktivitas, kadar air, serta karakter medium.

### **Input**

> Data yang diberikan dapat mencakup radargram, panjang atau posisi lintasan, waktu tempuh, frekuensi antena, skala kedalaman, dan informasi kondisi permukaan jika tersedia.

### **Constraints**

> Jangan menambahkan objek atau pola yang tidak terlihat pada data. Interpretasi harus berasal dari radargram dan informasi pendukung yang diberikan, dengan pemisahan yang jelas antara hasil pengamatan dan dugaan.

### **Output**

> Jelaskan reflektor yang tampak, tingkat kesinambungan lapisan, kemungkinan keberadaan objek atau batas material, perkiraan posisi atau kedalaman, serta karakter bawah permukaan yang ditunjukkan oleh pola radar.

### **Quality Criteria**

> Pembahasan harus memperhatikan skala waktu atau kedalaman, posisi lintasan, karakter refleksi, frekuensi antena, dan informasi tambahan. Gunakan istilah GPR secara sesuai dan hindari pernyataan yang terlalu pasti.

### **Verification**

> Pastikan interpretasi sesuai dengan pola pada radargram. Jika tersedia, cocokkan dengan kondisi lapangan, hasil pemboran atau penggalian, dan informasi dari metode lain.


### **4. STUDI KASUS GEOMAGNET**

### **Deskripsi**

Kajian ini menggunakan keluaran survei **geomagnet**, terutama peta medan magnet atau distribusi anomali magnetik.

Analisis ditujukan untuk mengenali variasi medan magnet yang dapat berkaitan dengan perbedaan sifat kemagnetan batuan, perubahan litologi, struktur, maupun objek tertentu di bawah permukaan.

### **Role**

> Anda bertindak sebagai pendamping analisis untuk membantu menjelaskan pola yang terlihat pada data geomagnet.

### **Task**

> Membantu membaca persebaran anomali magnetik dengan mempertimbangkan lokasi, bentuk, intensitas, dan perubahan polanya untuk memperoleh gambaran sumber anomali.

### **Context**

> Metode geomagnet mengamati perubahan medan magnet bumi yang berkaitan dengan sifat kemagnetan material. Kontras susceptibilitas magnetik dan kondisi geologi dapat menghasilkan variasi anomali pada suatu wilayah.

### **Input**

> Hasil dapat berupa peta anomali, nilai medan magnet, koordinat titik pengamatan, rentang warna, serta informasi geologi dan topografi yang relevan.

### **Constraints**

> Jangan menghasilkan nilai anomali baru atau mengubah data yang diberikan. Uraian harus berdasarkan pola yang tersedia dan membedakan antara anomali yang teramati, dugaan sumber, dan interpretasi yang masih memerlukan verifikasi.

### **Output**

> Uraikan sebaran anomali, perbedaan nilai, perubahan secara lateral, kemungkinan batas atau struktur, serta karakter sumber yang dapat dikaitkan dengan pola magnetik tersebut.

### **Quality Criteria**

> Pembacaan harus mengikuti legenda, skala, nilai anomali, koordinat, dan informasi pendukung. Gunakan konsep geomagnet secara benar dan jangan membuat kesimpulan yang tidak didukung oleh data.

### **Verification**

> Periksa kembali apakah interpretasi mengikuti pola anomali yang ditampilkan. Bila tersedia, bandingkan dengan kondisi geologi, topografi, pengamatan lapangan, atau hasil metode geofisika lain.


### **5. STUDI KASUS IP**

### **Deskripsi**

Kajian ini memanfaatkan hasil survei **Induced Polarization (IP)** yang dapat berupa penampang resistivitas dan distribusi chargeability.

Pembahasannya menitikberatkan pada hubungan pola resistivitas dan chargeability untuk membedakan zona atau material yang mempunyai respons polarisasi berbeda.

### **Role**

> Anda bertindak sebagai pendamping dalam membaca dan menerangkan hasil pengukuran metode IP.

### **Task**

> Membantu menginterpretasikan pola resistivitas dan chargeability sehingga perbedaan karakter material bawah permukaan dapat dijelaskan dari data yang tersedia.

### **Context**

> Metode IP mengukur respons material setelah diberi rangsangan arus listrik. Besarnya chargeability dapat berkaitan dengan komposisi material, kandungan mineral, porositas, fluida, serta karakter permukaan butir.

### **Input**

> Hasil dapat berupa penampang resistivitas, penampang chargeability, nilai parameter, konfigurasi dan posisi elektroda, kedalaman, serta informasi geologi pendukung.

### **Constraints**

> Jangan mengubah angka atau menciptakan data tambahan. Penafsiran harus bertumpu pada pola resistivitas dan chargeability yang diberikan serta memisahkan hasil pengamatan dari interpretasi dan kemungkinan penyebabnya.

### **Output**

> Jelaskan persebaran resistivitas dan chargeability, pembagian zona, perubahan vertikal maupun lateral, serta kemungkinan karakter material yang dapat menerangkan respons IP.

### **Quality Criteria**

> Interpretasi harus memperhatikan legenda, skala warna, rentang nilai kedua parameter, kedalaman, dan data pendukung. Hubungan antara resistivitas dan chargeability perlu dijelaskan secara hati-hati dengan istilah IP yang tepat.

### **Verification**

> Tinjau kembali kesesuaian interpretasi terhadap kedua penampang. Jika tersedia, gunakan data geologi, bor, analisis sampel, atau metode geofisika lain untuk membantu memeriksa hasil penafsiran.
